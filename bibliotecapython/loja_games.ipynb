```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression # biblioteca recomendada pela ia para utilizar a regressão linear 

# situação escolhida: A – Previsão de vendas

dados = { #tentei utilizar um arquivo que criei junto ao pd.read_csv() como feito nas aulas, mas não consegui, então criei um dicionário com os dados
    "ID_Produto": [1, 2, 3, 4, 5, 6, 7, 8],
    "Produto": ["Halo 3", "Pokémon Emerald", "Metal Gear Solid V", "LEGO Star Wars", "Resident Evil 4", "Balatro", "The Legend of Zeld: Breath of the Wild", "GTA VI"],
    "Qtd": [1, 2, 2, 3, 4, 5, 5, 6],
    "Preco_Unitario": [120.0, 250.0, 230.0, 360.0, 480.0, 590.0, 610.0, 720.0]
}

loja_games = pd.DataFrame(dados)

# variável de entrada (X) e variável-alvo (Y)
X = loja_games[["Qtd"]]
Y = loja_games["Preco_Unitario"]

# treinando o modelo de regressão linear
modelo = LinearRegression()
# padrão dos dados
modelo.fit(X, Y) 

# execução da previsão de um cliente que adicionou 8 produtos no carrinho, essa entrada é passada pro modelo.predict() e o resultado é armazenado na variável previsao_faturamento
carrinho_cliente = pd.DataFrame({"Qtd": [8]})
previsao_faturamento = modelo.predict(carrinho_cliente)


#resultado do modelo contendo custo fixo base, valor médio por produto e previsão de faturamento para 8 produtos
print("-- resultados do modelo --")
print(f"Custo Fixo Base: {modelo.intercept_:.2f}") # :.2f arredonda o float para duas casas decimais. pra deixar mais bonito na exibição
print(f"Valor Médio por Produto: {modelo.coef_[0]:.2f}") # modelo.coef é o atributo da biblioteca Scikit-Learn que guarda o coeficiente angular
print("--------------------------------" )
print(f"Previsão de Faturamento para 8 produtos: {previsao_faturamento[0]:.2f}")


# a loja opera sem qualquer taxa fixa de entrada ou frete (intercepto = 0), ou seja, o carrinho vazio não vai gerar custos 
# cada jogo adicionado ao carrinho gera um custo médio de R$ 120.00, e a previsão de faturamento para um cliente que adicionou 8 produtos no carrinho é de R$ 960,00
```
